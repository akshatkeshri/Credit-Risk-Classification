```python
import pandas as pd

base_path = "/content/drive/MyDrive/T2_Project_Akshat_Keshri/Dataset"

train_df = pd.read_csv(f"{base_path}/train_data.csv")
test_df = pd.read_csv(f"{base_path}/test_data.csv")
```

```python
X_train = train_df.drop('Risk', axis=1)
y_train = train_df['Risk']

X_test = test_df.drop('Risk', axis=1)
y_test = test_df['Risk']
```

Logistic Regression


```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
lr_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('classifier', LogisticRegression(random_state=42, max_iter=1000))
])
```

```python
lr_pipeline.fit(X_train, y_train)
```

```python
y_pred_lr = lr_pipeline.predict(X_test)
```

```python
lr_accuracy = accuracy_score(y_test, y_pred_lr)
lr_accuracy

confusion_matrix(y_test, y_pred_lr)

print(classification_report(y_test, y_pred_lr))
```

Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
rf_param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [10, 20, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}
```

```python
rf_grid = GridSearchCV(
    RandomForestClassifier(random_state=42),
    rf_param_grid,
    cv=5,
    scoring='f1',
    n_jobs=-1,
    verbose=1
)

rf_grid.fit(X_train, y_train)
```

```python
print("Best Random Forest Params:", rf_grid.best_params_)
print("Best CV F1 Score:", rf_grid.best_score_)
```

```python
rf_best = rf_grid.best_estimator_

y_pred_rf_tuned = rf_best.predict(X_test)

rf_tuned_accuracy = accuracy_score(y_test, y_pred_rf_tuned)
rf_tuned_accuracy

print(classification_report(y_test, y_pred_rf_tuned))
```

```python
from xgboost import XGBClassifier
```

```python
xgb_param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.05, 0.1],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}
```

```python
xgb_grid = GridSearchCV(
    XGBClassifier(
        random_state=42,
        eval_metric='logloss',
        use_label_encoder=False
    ),
    xgb_param_grid,
    cv=5,
    scoring='f1',
    n_jobs=-1,
    verbose=1
)

xgb_grid.fit(X_train, y_train)
```

```python
print("Best XGBoost Params:", xgb_grid.best_params_)
print("Best CV F1 Score:", xgb_grid.best_score_)
```

```python
xgb_best = xgb_grid.best_estimator_

y_pred_xgb_tuned = xgb_best.predict(X_test)

xgb_tuned_accuracy = accuracy_score(y_test, y_pred_xgb_tuned)
xgb_tuned_accuracy

print(classification_report(y_test, y_pred_xgb_tuned))
```

```python
import pandas as pd

comparison = pd.DataFrame({
    'Model': ['Random Forest (Baseline)', 'Random Forest (Tuned)',
              'XGBoost (Baseline)', 'XGBoost (Tuned)'],
    'Accuracy': [
        rf_accuracy,
        rf_tuned_accuracy,
        xgb_accuracy,
        xgb_tuned_accuracy
    ]
})

comparison
```

Experiment Tracking with MLflow

```python
!pip install mlflow

import mlflow
import mlflow.sklearn

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
mlflow.set_experiment("credit_risk_classification")
```

```python
def log_confusion_matrix(y_true, y_pred, run_name):
    cm = confusion_matrix(y_true, y_pred)

    plt.figure(figsize=(4,3))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.title("Confusion Matrix")

    file_name = f"{run_name}_confusion_matrix.png"
    plt.savefig(file_name, bbox_inches='tight')
    plt.close()

    mlflow.log_artifact(file_name)
    os.remove(file_name)
```

```python
with mlflow.start_run(run_name="RandomForest_Tuned"):

    mlflow.log_param("model_type", "Random Forest")
    mlflow.log_params(rf_grid.best_params_)

    y_pred_rf_tuned = rf_best.predict(X_test)

    acc = accuracy_score(y_test, y_pred_rf_tuned)
    prec = precision_score(y_test, y_pred_rf_tuned)
    rec = recall_score(y_test, y_pred_rf_tuned)
    f1 = f1_score(y_test, y_pred_rf_tuned)

    mlflow.log_metric("accuracy", acc)
    mlflow.log_metric("precision", prec)
    mlflow.log_metric("recall", rec)
    mlflow.log_metric("f1_score", f1)

    log_confusion_matrix(y_test, y_pred_rf_tuned, "rf_tuned")

```

```python
with mlflow.start_run(run_name="XGBoost_Tuned"):

    mlflow.log_param("model_type", "XGBoost")
    mlflow.log_params(xgb_grid.best_params_)

    y_pred_xgb_tuned = xgb_best.predict(X_test)

    acc = accuracy_score(y_test, y_pred_xgb_tuned)
    prec = precision_score(y_test, y_pred_xgb_tuned)
    rec = recall_score(y_test, y_pred_xgb_tuned)
    f1 = f1_score(y_test, y_pred_xgb_tuned)

    mlflow.log_metric("accuracy", acc)
    mlflow.log_metric("precision", prec)
    mlflow.log_metric("recall", rec)
    mlflow.log_metric("f1_score", f1)

    log_confusion_matrix(y_test, y_pred_xgb_tuned, "xgb_tuned")

```

```python
mlflow.search_runs(experiment_names=["credit_risk_classification"])
```
